### Завдання 25


Додатково знайдіть сам кут у градусах.

````python
a = np.array([1, 2])
b = np.array([2, 1])

cos_angle = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
angle_degrees = np.degrees(np.arccos(cos_angle))

print("Косинус кута:", cos_angle)
print("Кут у градусах:", angle_degrees)
````

### Завдання 28
Для матриці:

```text
A = [[4, 7],
     [2, 6]]
```

обчисліть:
- визначник;
- обернену матрицю.

Перевірте результат, помноживши початкову матрицю на обернену.

````python
A = np.array([
    [4, 7],
    [2, 6]
])

det = np.linalg.det(A)
A_inv = np.linalg.inv(A)
check = A @ A_inv

print("Визначник:", det)
print("Обернена матриця:")
print(A_inv)
print("A @ A_inv:")
print(check)
print("Чи близький результат до одиничної матриці:",
      np.allclose(check, np.eye(2)))
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

mean = np.mean(A, axis=0)
std = np.std(A, axis=0)

Z = (A - mean) / std

print("Середні значення стовпців:", mean)
print("Стандартні відхилення стовпців:", std)
print("Нормалізована матриця:")
print(Z)
````

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

````python
sales = np.array([
    [120, 150, 170, 160],
    [80,  90,  100, 110],
    [200, 180, 210, 230]
])

product_totals = np.sum(sales, axis=1)
month_totals = np.sum(sales, axis=0)
best_product_index = np.argmax(product_totals)
best_month_index = np.argmax(month_totals)
product_averages = np.mean(sales, axis=1)

print("Сумарні продажі кожного товару:", product_totals)
print("Сумарні продажі кожного місяця:", month_totals)
print("Індекс товару з найбільшими продажами:", best_product_index)
print("Номер місяця з найбільшими продажами:", best_month_index + 1)
print("Середні продажі кожного товару:", product_averages)
````

Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
np.random.seed(42)
A = np.random.randint(1, 21, size=(5, 5))

main_diag_sum = np.trace(A)
secondary_diag_sum = np.trace(np.fliplr(A))
row_max = np.max(A, axis=1)
column_min = np.min(A, axis=0)
mean_value = np.mean(A)

result = A.copy()
result[result < mean_value] = 0

print("Матриця:")
print(A)
print("Сума головної діагоналі:", main_diag_sum)
print("Сума побічної діагоналі:", secondary_diag_sum)
print("Максимуми рядків:", row_max)
print("Мінімуми стовпців:", column_min)
print("Середнє значення:", mean_value)
print("Матриця після заміни елементів < середнього на 0:")
print(result)
````

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
A = np.array([
    [1, 2, 1],
    [2, 1, -1],
    [3, -1, 2]
])
b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)
Ax = A @ x

print("Розв'язок [x, y, z]:", x)
print("A @ x:", Ax)
print("b:", b)
print("Чи збігаються A @ x та b:", np.allclose(Ax, b))
````
